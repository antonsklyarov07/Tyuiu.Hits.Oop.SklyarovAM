```
using System;

interface IReturnable
{
    void ReturnProduct();
}

interface IGiftWrappable
{
    void GiftWrap();
}

class Product
{
    private string _name;
    private double _price;
    private string _manufacturer;
    private int _stock;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public double Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
                _price = value;
            else
                Console.WriteLine("Ошибка: цена не может быть отрицательной!");
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = value; }
    }

    public int Stock
    {
        get { return _stock; }
        set
        {
            if (value >= 0)
                _stock = value;
            else
                Console.WriteLine("Ошибка: количество на складе не может быть отрицательным!");
        }
    }

    public Product(string name, double price, string manufacturer, int stock)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Stock = stock;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, На складе: {Stock} шт.";
    }

    public virtual void Discount(double percentage)
    {
        Price = Price - (Price * percentage / 100);
        Console.WriteLine($"Применена скидка {percentage}%. Новая цена: {Price} руб.");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public void CompareWith(Product other)
    {
        Console.WriteLine($"\nСравнение: {Name} и {other.Name}");
        if (Price > other.Price)
            Console.WriteLine($"{Name} дороже на {Price - other.Price} руб.");
        else if (Price < other.Price)
            Console.WriteLine($"{other.Name} дороже на {other.Price - Price} руб.");
        else
            Console.WriteLine("Цены одинаковые");
    }

    public void Sell(int quantity)
    {
        if (quantity <= Stock)
        {
            Stock -= quantity;
            Console.WriteLine($"Продано {quantity} шт. товара {Name}. Осталось на складе: {Stock}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {Name} на складе!");
        }
    }

    public void Restock(int quantity)
    {
        Stock += quantity;
        Console.WriteLine($"Пополнение склада: {Name} +{quantity} шт. Теперь на складе: {Stock}");
    }
}

class Electronics : Product, IGiftWrappable
{
    private int _warrantyPeriod;
    private int _powerConsumption;
    private bool _isSmart;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set
        {
            if (value >= 0)
                _warrantyPeriod = value;
            else
                Console.WriteLine("Ошибка: гарантия не может быть отрицательной!");
        }
    }

    public int PowerConsumption
    {
        get { return _powerConsumption; }
        set { _powerConsumption = value; }
    }

    public bool IsSmart
    {
        get { return _isSmart; }
        set { _isSmart = value; }
    }

    public Electronics(string name, double price, string manufacturer, int stock,
                       int warrantyPeriod, int powerConsumption, bool isSmart)
        : base(name, price, manufacturer, stock)
    {
        WarrantyPeriod = warrantyPeriod;
        PowerConsumption = powerConsumption;
        IsSmart = isSmart;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес., Потребление: {PowerConsumption} Вт, Умное устройство: {IsSmart}";
    }

    public override void Discount(double percentage)
    {
        if (WarrantyPeriod > 12)
        {
            percentage += 5;
            Console.WriteLine($"Дополнительная скидка 5% за гарантию более 12 месяцев");
        }
        base.Discount(percentage);
    }

    public void ExtendWarranty(int months)
    {
        WarrantyPeriod += months;
        Console.WriteLine($"Гарантия продлена на {months} мес. Теперь: {WarrantyPeriod} мес.");
    }

    public void CheckWarranty()
    {
        Console.WriteLine(WarrantyPeriod > 0
            ? $"На {Name} действует гарантия {WarrantyPeriod} мес."
            : $"Гарантия на {Name} истекла.");
    }

    public void GiftWrap()
    {
        Console.WriteLine($"{Name} упакован в подарочную коробку.");
    }
}

class Smartphone : Electronics
{
    private string _operatingSystem;
    private int _cameraMegapixels;

    public string OperatingSystem
    {
        get { return _operatingSystem; }
        set { _operatingSystem = value; }
    }

    public int CameraMegapixels
    {
        get { return _cameraMegapixels; }
        set { _cameraMegapixels = value; }
    }

    public Smartphone(string name, double price, string manufacturer, int stock,
                      int warrantyPeriod, int powerConsumption, bool isSmart,
                      string operatingSystem, int cameraMegapixels)
        : base(name, price, manufacturer, stock, warrantyPeriod, powerConsumption, isSmart)
    {
        OperatingSystem = operatingSystem;
        CameraMegapixels = cameraMegapixels;
    }

    public void TakePhoto()
    {
        Console.WriteLine($"{Name} делает фото на {CameraMegapixels} Мп камеру!");
    }

    public void InstallApp(string appName)
    {
        Console.WriteLine($"Приложение {appName} установлено на {Name} ({OperatingSystem}).");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", ОС: {OperatingSystem}, Камера: {CameraMegapixels} Мп";
    }
}

class SmartWatch : Electronics, IReturnable
{
    private bool _isWaterproof;
    private int _batteryCapacity;

    public bool IsWaterproof
    {
        get { return _isWaterproof; }
        set { _isWaterproof = value; }
    }

    public int BatteryCapacity
    {
        get { return _batteryCapacity; }
        set { _batteryCapacity = value; }
    }

    public SmartWatch(string name, double price, string manufacturer, int stock,
                      int warrantyPeriod, int powerConsumption, bool isSmart,
                      bool isWaterproof, int batteryCapacity)
        : base(name, price, manufacturer, stock, warrantyPeriod, powerConsumption, isSmart)
    {
        IsWaterproof = isWaterproof;
        BatteryCapacity = batteryCapacity;
    }

    public void TrackHeartRate()
    {
        Console.WriteLine($"{Name} измеряет пульс...");
    }

    public void ReturnProduct()
    {
        Console.WriteLine($"{Name} возвращён в магазин. Деньги возвращены покупателю.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Влагозащита: {IsWaterproof}, Батарея: {BatteryCapacity} мАч";
    }
}

class Clothing : Product, IReturnable
{
    private string _size;
    private string _color;
    private string _material;

    public string Size
    {
        get { return _size; }
        set { _size = value; }
    }

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public string Material
    {
        get { return _material; }
        set { _material = value; }
    }

    public Clothing(string name, double price, string manufacturer, int stock,
                    string size, string color, string material)
        : base(name, price, manufacturer, stock)
    {
        Size = size;
        Color = color;
        Material = material;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}, Цвет: {Color}, Материал: {Material}";
    }

    public override void Display()
    {
        Console.WriteLine($"Одежда: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, Размер: {Size}, Цвет: {Color}");
    }

    public void TryOn(Product other)
    {
        Console.WriteLine($"\nПримерка: {Name} (размер {Size}) вместе с {other.Name}");
    }

    public void Wash()
    {
        Console.WriteLine($"{Name} постиран. Состав: {Material}.");
    }

    public bool IsSizeAvailable(string size)
    {
        return Size == size;
    }

    public void ReturnProduct()
    {
        Console.WriteLine($"{Name} возвращён в магазин. Товар проверен и снова на витрине.");
    }
}

class Books : Product
{
    private string _author;
    private int _pageCount;
    private int _year;

    public string Author
    {
        get { return _author; }
        set { _author = value; }
    }

    public int PageCount
    {
        get { return _pageCount; }
        set { _pageCount = value; }
    }

    public int Year
    {
        get { return _year; }
        set { _year = value; }
    }

    public Books(string name, double price, string manufacturer, int stock,
                 string author, int pageCount, int year)
        : base(name, price, manufacturer, stock)
    {
        Author = author;
        PageCount = pageCount;
        Year = year;
    }

    public override string GetInfo()
    {
        return $"Название: {Name}, Автор: {Author}, Цена: {Price} руб., Издатель: {Manufacturer}, Страниц: {PageCount}, Год: {Year}";
    }

    public void ReadTogether(Product other)
    {
        Console.WriteLine($"\nЧтение: {Name} ({Author}) и {other.Name}");
    }

    public void Read(int pages)
    {
        Console.WriteLine($"Прочитано {pages} страниц из {PageCount}. Осталось: {PageCount - pages}.");
    }

    public bool IsBestseller()
    {
        return Year >= 2020;
    }
}

class Program
{
    static void Main()
    {
        Electronics laptop = new Electronics("Ноутбук", 50000, "HP", 10, 24, 65, true);
        Clothing tshirt = new Clothing("Футболка", 1500, "Nike", 25, "L", "Белый", "Хлопок");
        Books book = new Books("Война и мир", 800, "Эксмо", 30, "Лев Толстой", 1300, 2021);
        Smartphone phone = new Smartphone("Galaxy S23", 70000, "Samsung", 15, 24, 25, true, "Android", 50);
        SmartWatch watch = new SmartWatch("Apple Watch", 35000, "Apple", 8, 12, 5, true, true, 300);

        Console.WriteLine("=== Информация о продуктах ===");
        laptop.Display();
        tshirt.Display();
        book.Display();
        phone.Display();
        watch.Display();

        Console.WriteLine("\n=== Применение скидок ===");
        laptop.Discount(10);
        tshirt.Discount(10);
        book.Discount(10);

        Console.WriteLine("\n=== Обновленная информация ===");
        laptop.Display();
        tshirt.Display();
        book.Display();

        Console.WriteLine("\n=== Взаимодействие объектов ===");
        laptop.CompareWith(book);
        tshirt.TryOn(book);
        book.ReadTogether(laptop);

        Console.WriteLine("\n=== Проверка валидации ===");
        laptop.Price = -100;
        laptop.WarrantyPeriod = -5;

        Console.WriteLine("\n=== Дополнительно ===");
        laptop.ExtendWarranty(12);
        laptop.CheckWarranty();
        laptop.GiftWrap();
        laptop.Discount(10);

        Console.WriteLine("\n=== Смартфон (сложное наследование) ===");
        phone.TakePhoto();
        phone.InstallApp("Telegram");
        phone.ExtendWarranty(6);
        phone.GiftWrap();
        phone.Sell(3);

        Console.WriteLine("\n=== Смарт-часы (сложное + множественное наследование) ===");
        watch.TrackHeartRate();
        watch.GiftWrap();
        watch.ReturnProduct();
        watch.CheckWarranty();

        Console.WriteLine("\n=== Одежда и книги ===");
        tshirt.Wash();
        Console.WriteLine(tshirt.IsSizeAvailable("L") ? "Размер L есть в наличии" : "Размера L нет");
        tshirt.ReturnProduct();

        book.Read(120);
        Console.WriteLine(book.IsBestseller() ? "Книга — бестселлер!" : "Обычная книга");
        book.Restock(10);
    }
}
```
