```
using System;
using System.Collections.Generic;
using System.Linq;

Electronics laptop = new Electronics("Ноутбук", 50000, "HP", 10, 24, 65, true);
Clothing tshirt = new Clothing("Футболка", 1500, "Nike", 25, "L", "Белый", "Хлопок");
Books book = new Books("Война и мир", 800, "Эксмо", 30, "Лев Толстой", 1300, 2021);
Smartphone phone = new Smartphone("Galaxy S23", 70000, "Samsung", 15, 24, 25, true, "Android", 50);
SmartWatch watch = new SmartWatch("Apple Watch", 35000, "Apple", 8, 12, 5, true, true, 300);

laptop.SKU = 1001; laptop.Category = "Электроника"; laptop.Rating = 4.5; laptop.Supplier = "HP Russia"; laptop.DiscountLimit = 30; laptop.IsFeatured = true;
tshirt.SKU = 1002; tshirt.Category = "Одежда"; tshirt.Rating = 4.8; tshirt.Supplier = "Nike Official"; tshirt.DiscountLimit = 50; tshirt.IsFeatured = false;
book.SKU = 1003; book.Category = "Книги"; book.Rating = 4.9; book.Supplier = "Эксмо-Пресс"; book.DiscountLimit = 25; book.IsFeatured = true;
phone.SKU = 1004; phone.Category = "Смартфоны"; phone.Rating = 4.7; phone.Supplier = "Samsung CIS"; phone.DiscountLimit = 20; phone.IsFeatured = true;
watch.SKU = 1005; watch.Category = "Гаджеты"; watch.Rating = 4.6; watch.Supplier = "Apple Store"; watch.DiscountLimit = 15; watch.IsFeatured = false;

Console.WriteLine("=== 1. События и лямбда-подписки ===");
List<Product> allProducts = new List<Product> { laptop, tshirt, book, phone, watch };

foreach (var item in allProducts)
{
    item.PriceChanged += (name, oldPrice, newPrice) =>
        Console.WriteLine($"[СОБЫТИЕ ЦЕНЫ] Товар «{name}»: {oldPrice} руб. -> {newPrice} руб.");

    item.StockLow += (name, currentStock) =>
        Console.WriteLine($"[ВНИМАНИЕ СКЛАД] Низкий остаток товара «{name}»: всего {currentStock} шт.!");
}

laptop.Discount(10);
watch.Sell(6, "Алексей");

Console.WriteLine("\n=== 2. Делегаты ===");
ProductActionDelegate batchAction = laptop.Display;
batchAction += tshirt.Display;
batchAction += book.Display;
batchAction();

Console.WriteLine("\n=== 3. Коллекции: List, Dictionary, HashSet ===");
Dictionary<int, Product> catalogBySku = new Dictionary<int, Product>();
foreach (var item in allProducts)
{
    catalogBySku[item.SKU] = item;
}

if (catalogBySku.TryGetValue(1004, out Product foundBySku))
{
    Console.WriteLine("Найден товар по SKU 1004:");
    foundBySku.Display();
}

HashSet<Product> uniqueCart = new HashSet<Product>();
Console.WriteLine($"\nДобавление ноутбука в HashSet: {uniqueCart.Add(laptop)}");
Console.WriteLine($"Повторное добавление того же ноутбука в HashSet: {uniqueCart.Add(laptop)}");
Console.WriteLine($"Добавление книги в HashSet: {uniqueCart.Add(book)}");
Console.WriteLine($"Уникальных товаров в корзине: {uniqueCart.Count}");

Console.WriteLine("\n=== 4. Лямбда-выражения и LINQ ===");
var expensiveFeatured = allProducts
    .Where(p => p.IsFeatured && p.Price > 10000)
    .OrderByDescending(p => p.Price)
    .ToList();

Console.WriteLine("Рекомендуемые товары дороже 10 000 руб.:");
expensiveFeatured.ForEach(p => Console.WriteLine($"- {p.Name}: {p.Price} руб. (Рейтинг: {p.Rating})"));

Console.WriteLine("\n=== 5. Новые атрибуты и методы классов ===");
laptop.UpdateSupplier("ООО ТехноИмпорт");
laptop.ToggleFeatured();
Console.WriteLine($"Налог (20%) на {laptop.Name}: {laptop.CalculateTax(20)} руб.");

laptop.SerialNumber = "SN-HP-998877";
laptop.OperatingVoltage = 220;
laptop.HasBattery = true;
laptop.RunDiagnostics();
laptop.CheckVoltageCompatibility(220);
Console.WriteLine($"Нужна утилизация батареи: {laptop.RequiresBatteryDisposal()}");

phone.ScreenSize = 6.1;
phone.RefreshRate = 120;
phone.Has5G = true;
phone.CheckNetworkSupport();
phone.SetRefreshRate(60);
phone.CompareCamera(48);

watch.StrapMaterial = "Фторэластомер";
watch.WaterResistanceMeters = 50;
watch.StepCount = 8450;
watch.CheckSwimmingSuitability();
watch.ResetSteps();
watch.ChangeStrap("Миланский сетчатый браслет");

tshirt.CountryOfOrigin = "Вьетнам";
tshirt.WashTemperature = 40;
tshirt.Gender = "Унисекс";
tshirt.PrintCareLabel();
tshirt.CheckGenderFit("Унисекс");
tshirt.ApplySeasonalMarkup(10);

book.CoverType = "Твердый переплет";
book.Language = "Русский";
book.AgeRestriction = 12;
book.CheckAgeAccess(16);
book.TranslateInfo("Английский");
book.CalculateWeightByPages(1.5);

Console.WriteLine("\n=== 6. Явная реализация интерфейсов и DI ===");
((IReturnable)watch).ReturnProduct();
((IReturnable)tshirt).ReturnProduct();

ProductService service = new ProductService(phone);
service.ShowProductCard();
service.CheckAvailability();

public delegate void ProductActionDelegate();
public delegate void PriceChangedHandler(string productName, double oldPrice, double newPrice);
public delegate void StockLowHandler(string productName, int remainingStock);

interface IReturnable
{
    void ReturnProduct();
}

interface IGiftWrappable
{
    void GiftWrap();
}

interface IProduct
{
    string Name { get; set; }
    double Price { get; set; }
    string Manufacturer { get; set; }
    int Stock { get; set; }
    void Display();
    void Discount(double percentage);
    string GetInfo();
}

class Product : IProduct
{
    private string _name;
    private double _price;
    private string _manufacturer;
    private int _stock;
    private int _sku;
    private string _category;
    private double _rating;
    private string _supplier;
    private double _discountLimit;
    private bool _isFeatured;

    public event PriceChangedHandler PriceChanged;
    public event StockLowHandler StockLow;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public double Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
            {
                double oldPrice = _price;
                _price = value;
                if (oldPrice > 0 && oldPrice != _price)
                {
                    PriceChanged?.Invoke(Name, oldPrice, _price);
                }
            }
            else
                Console.WriteLine("Ошибка: цена не может быть отрицательной!");
        }
    }

    public string Manufacturer
    {
        get { return _manufacturer; }
        set { _manufacturer = value; }
    }

    public int Stock
    {
        get { return _stock; }
        set
        {
            if (value >= 0)
            {
                _stock = value;
                if (_stock <= 3)
                {
                    StockLow?.Invoke(Name, _stock);
                }
            }
            else
                Console.WriteLine("Ошибка: количество на складе не может быть отрицательным!");
        }
    }

    public int SKU
    {
        get { return _sku; }
        set
        {
            if (value >= 0)
                _sku = value;
            else
                Console.WriteLine("Ошибка: артикул не может быть отрицательным!");
        }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }

    public double Rating
    {
        get { return _rating; }
        set
        {
            if (value >= 0 && value <= 5)
                _rating = value;
            else
                Console.WriteLine("Ошибка: рейтинг должен быть от 0 до 5!");
        }
    }

    public string Supplier
    {
        get { return _supplier; }
        set { _supplier = value; }
    }

    public double DiscountLimit
    {
        get { return _discountLimit; }
        set { _discountLimit = value >= 0 && value <= 100 ? value : 50; }
    }

    public bool IsFeatured
    {
        get { return _isFeatured; }
        set { _isFeatured = value; }
    }

    public Product(string name, double price, string manufacturer, int stock)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Stock = stock;
        DiscountLimit = 50;
    }

    public override bool Equals(object obj)
    {
        if (obj is Product other)
        {
            return Name == other.Name && Manufacturer == other.Manufacturer && SKU == other.SKU;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Manufacturer, SKU);
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, На складе: {Stock} шт.";
    }

    public virtual void Discount(double percentage)
    {
        if (percentage > DiscountLimit)
        {
            percentage = DiscountLimit;
        }
        Price = Price - (Price * percentage / 100);
        Console.WriteLine($"Применена скидка {percentage}%. Новая цена: {Price} руб.");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public void CompareWith(Product other)
    {
        Console.WriteLine($"\nСравнение: {Name} и {other.Name}");
        if (Price > other.Price)
            Console.WriteLine($"{Name} дороже на {Price - other.Price} руб.");
        else if (Price < other.Price)
            Console.WriteLine($"{other.Name} дороже на {other.Price - Price} руб.");
        else
            Console.WriteLine("Цены одинаковые");
    }

    public void Sell(int quantity, string customer)
    {
        if (quantity <= Stock)
        {
            Stock -= quantity;
            Console.WriteLine($"{customer} купил {quantity} шт. товара {Name}. Осталось: {Stock}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара {Name} на складе!");
        }
    }

    public void Sell(int quantity)
    {
        Sell(quantity, "Анонимный покупатель");
    }

    public void Restock(int quantity)
    {
        Stock += quantity;
        Console.WriteLine($"Пополнение склада: {Name} +{quantity} шт. Теперь на складе: {Stock}");
    }

    public void AddRating(double stars)
    {
        if (stars >= 0 && stars <= 5)
        {
            Rating = (Rating + stars) / 2;
            Console.WriteLine($"Новый рейтинг {Name}: {Rating:F1} из 5");
        }
    }

    public bool IsAvailable()
    {
        return Stock > 0;
    }

    public void PrintSKU()
    {
        Console.WriteLine($"Артикул {Name}: {SKU}, категория: {Category}");
    }

    public void UpdateSupplier(string newSupplier)
    {
        Supplier = newSupplier;
        Console.WriteLine($"Поставщик товара {Name} обновлен: {Supplier}");
    }

    public void ToggleFeatured()
    {
        IsFeatured = !IsFeatured;
        Console.WriteLine($"Статус рекомендации для {Name}: {(IsFeatured ? "Рекомендуемый" : "Обычный")}");
    }

    public double CalculateTax(double taxRate)
    {
        return Price * (taxRate / 100.0);
    }
}

class Electronics : Product, IGiftWrappable
{
    private int _warrantyPeriod;
    private int _powerConsumption;
    private bool _isSmart;
    private string _energyClass;
    private double _weight;
    private string _serialNumber;
    private int _operatingVoltage;
    private bool _hasBattery;

    public int WarrantyPeriod
    {
        get { return _warrantyPeriod; }
        set
        {
            if (value >= 0)
                _warrantyPeriod = value;
            else
                Console.WriteLine("Ошибка: гарантия не может быть отрицательной!");
        }
    }

    public int PowerConsumption
    {
        get { return _powerConsumption; }
        set { _powerConsumption = value; }
    }

    public bool IsSmart
    {
        get { return _isSmart; }
        set { _isSmart = value; }
    }

    public string EnergyClass
    {
        get { return _energyClass; }
        set { _energyClass = value; }
    }

    public double Weight
    {
        get { return _weight; }
        set
        {
            if (value >= 0)
                _weight = value;
            else
                Console.WriteLine("Ошибка: вес не может быть отрицательным!");
        }
    }

    public string SerialNumber
    {
        get { return _serialNumber; }
        set { _serialNumber = value; }
    }

    public int OperatingVoltage
    {
        get { return _operatingVoltage; }
        set { _operatingVoltage = value; }
    }

    public bool HasBattery
    {
        get { return _hasBattery; }
        set { _hasBattery = value; }
    }

    public Electronics(string name, double price, string manufacturer, int stock,
                       int warrantyPeriod, int powerConsumption, bool isSmart)
        : base(name, price, manufacturer, stock)
    {
        WarrantyPeriod = warrantyPeriod;
        PowerConsumption = powerConsumption;
        IsSmart = isSmart;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес., Потребление: {PowerConsumption} Вт, Умное устройство: {IsSmart}";
    }

    public override void Discount(double percentage)
    {
        if (WarrantyPeriod > 12)
        {
            percentage += 5;
            Console.WriteLine("Дополнительная скидка 5% за гарантию более 12 месяцев");
        }
        base.Discount(percentage);
    }

    public void ExtendWarranty(int months)
    {
        WarrantyPeriod += months;
        Console.WriteLine($"Гарантия продлена на {months} мес. Теперь: {WarrantyPeriod} мес.");
    }

    public void CheckWarranty()
    {
        Console.WriteLine(WarrantyPeriod > 0
            ? $"На {Name} действует гарантия {WarrantyPeriod} мес."
            : $"Гарантия на {Name} истекла.");
    }

    public void GiftWrap()
    {
        Console.WriteLine($"{Name} упакован в подарочную коробку.");
    }

    public void PrintSpecs()
    {
        Console.WriteLine($"{Name}: класс энергоэффективности {EnergyClass}, вес {Weight} кг");
    }

    public double CalcYearlyPowerCost(double pricePerKwh)
    {
        double cost = PowerConsumption / 1000.0 * 24 * 365 * pricePerKwh;
        Console.WriteLine($"Годовое энергопотребление {Name}: {cost:F2} руб. (при {pricePerKwh} руб/кВт·ч)");
        return cost;
    }

    public void RunDiagnostics()
    {
        Console.WriteLine($"Диагностика {Name} (S/N: {SerialNumber}): ошибок не обнаружено.");
    }

    public void CheckVoltageCompatibility(int gridVoltage)
    {
        Console.WriteLine(OperatingVoltage == gridVoltage
            ? $"{Name} совместим с сетью {gridVoltage} В."
            : $"{Name} требует адаптер для сети {gridVoltage} В.");
    }

    public bool RequiresBatteryDisposal()
    {
        return HasBattery;
    }
}

class Smartphone : Electronics
{
    private string _operatingSystem;
    private int _cameraMegapixels;
    private int _storageGB;
    private int _batteryMAh;
    private double _screenSize;
    private int _refreshRate;
    private bool _has5G;

    public string OperatingSystem
    {
        get { return _operatingSystem; }
        set { _operatingSystem = value; }
    }

    public int CameraMegapixels
    {
        get { return _cameraMegapixels; }
        set { _cameraMegapixels = value; }
    }

    public int StorageGB
    {
        get { return _storageGB; }
        set { _storageGB = value; }
    }

    public int BatteryMAh
    {
        get { return _batteryMAh; }
        set { _batteryMAh = value; }
    }

    public double ScreenSize
    {
        get { return _screenSize; }
        set { _screenSize = value; }
    }

    public int RefreshRate
    {
        get { return _refreshRate; }
        set { _refreshRate = value; }
    }

    public bool Has5G
    {
        get { return _has5G; }
        set { _has5G = value; }
    }

    public Smartphone(string name, double price, string manufacturer, int stock,
                      int warrantyPeriod, int powerConsumption, bool isSmart,
                      string operatingSystem, int cameraMegapixels)
        : base(name, price, manufacturer, stock, warrantyPeriod, powerConsumption, isSmart)
    {
        OperatingSystem = operatingSystem;
        CameraMegapixels = cameraMegapixels;
    }

    public void TakePhoto()
    {
        Console.WriteLine($"{Name} делает фото на {CameraMegapixels} Мп камеру!");
    }

    public void InstallApp(string appName)
    {
        Console.WriteLine($"Приложение {appName} установлено на {Name} ({OperatingSystem}).");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", ОС: {OperatingSystem}, Камера: {CameraMegapixels} Мп";
    }

    public void FormatStorage()
    {
        Console.WriteLine($"Память {StorageGB} ГБ на {Name} очищена.");
    }

    public void Charge()
    {
        Console.WriteLine($"{Name} заряжается... {BatteryMAh} мАч, время полной зарядки ~{BatteryMAh / 1000.0:F1} ч");
    }

    public void CheckNetworkSupport()
    {
        Console.WriteLine($"{Name} (экран {ScreenSize}\"): поддержка 5G — {(Has5G ? "Есть" : "Нет")}.");
    }

    public void SetRefreshRate(int hz)
    {
        RefreshRate = hz;
        Console.WriteLine($"Частота экрана {Name} установлена на {RefreshRate} Гц.");
    }

    public void CompareCamera(int targetMp)
    {
        Console.WriteLine(CameraMegapixels >= targetMp
            ? $"Камера {Name} ({CameraMegapixels} Мп) соответствует требованию ({targetMp} Мп)."
            : $"Камера {Name} слабее требуемых {targetMp} Мп.");
    }
}

class SmartWatch : Electronics, IReturnable
{
    private bool _isWaterproof;
    private int _batteryCapacity;
    private bool _hasGPS;
    private int _heartRate;
    private string _strapMaterial;
    private int _waterResistanceMeters;
    private int _stepCount;

    public bool IsWaterproof
    {
        get { return _isWaterproof; }
        set { _isWaterproof = value; }
    }

    public int BatteryCapacity
    {
        get { return _batteryCapacity; }
        set { _batteryCapacity = value; }
    }

    public bool HasGPS
    {
        get { return _hasGPS; }
        set { _hasGPS = value; }
    }

    public int HeartRate
    {
        get { return _heartRate; }
        set { _heartRate = value; }
    }

    public string StrapMaterial
    {
        get { return _strapMaterial; }
        set { _strapMaterial = value; }
    }

    public int WaterResistanceMeters
    {
        get { return _waterResistanceMeters; }
        set { _waterResistanceMeters = value; }
    }

    public int StepCount
    {
        get { return _stepCount; }
        set { _stepCount = value; }
    }

    public SmartWatch(string name, double price, string manufacturer, int stock,
                      int warrantyPeriod, int powerConsumption, bool isSmart,
                      bool isWaterproof, int batteryCapacity)
        : base(name, price, manufacturer, stock, warrantyPeriod, powerConsumption, isSmart)
    {
        IsWaterproof = isWaterproof;
        BatteryCapacity = batteryCapacity;
    }

    public void TrackHeartRate()
    {
        Console.WriteLine($"{Name} измеряет пульс...");
    }

    void IReturnable.ReturnProduct()
    {
        Console.WriteLine($"{Name} возвращён в магазин. Часы проверены и сброшены до заводских настроек. Деньги возвращены покупателю.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Влагозащита: {IsWaterproof}, Батарея: {BatteryCapacity} мАч";
    }

    public void StartWorkout(string type)
    {
        Console.WriteLine($"{Name}: начата тренировка «{type}». GPS: {(HasGPS ? "вкл" : "выкл")}");
    }

    public void MeasureHeartRate()
    {
        Random rnd = new Random();
        HeartRate = rnd.Next(60, 120);
        Console.WriteLine($"{Name}: пульс {HeartRate} уд/мин.");
    }

    public void CheckSwimmingSuitability()
    {
        Console.WriteLine(WaterResistanceMeters >= 50
            ? $"{Name} ({WaterResistanceMeters} м) подходят для плавания в бассейне."
            : $"{Name} не предназначены для глубокого погружения.");
    }

    public void ResetSteps()
    {
        Console.WriteLine($"Счетчик шагов {Name} (было: {StepCount}) сброшен на 0.");
        StepCount = 0;
    }

    public void ChangeStrap(string newMaterial)
    {
        StrapMaterial = newMaterial;
        Console.WriteLine($"На {Name} установлен новый ремешок: {StrapMaterial}.");
    }
}

class Clothing : Product, IReturnable
{
    private string _size;
    private string _color;
    private string _material;
    private string _season;
    private bool _isOnSale;
    private string _countryOfOrigin;
    private int _washTemperature;
    private string _gender;

    public string Size
    {
        get { return _size; }
        set { _size = value; }
    }

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public string Material
    {
        get { return _material; }
        set { _material = value; }
    }

    public string Season
    {
        get { return _season; }
        set { _season = value; }
    }

    public bool IsOnSale
    {
        get { return _isOnSale; }
        set { _isOnSale = value; }
    }

    public string CountryOfOrigin
    {
        get { return _countryOfOrigin; }
        set { _countryOfOrigin = value; }
    }

    public int WashTemperature
    {
        get { return _washTemperature; }
        set { _washTemperature = value; }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public Clothing(string name, double price, string manufacturer, int stock,
                    string size, string color, string material)
        : base(name, price, manufacturer, stock)
    {
        Size = size;
        Color = color;
        Material = material;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Размер: {Size}, Цвет: {Color}, Материал: {Material}";
    }

    public override void Display()
    {
        Console.WriteLine($"Одежда: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, Размер: {Size}, Цвет: {Color}");
    }

    public void TryOn(Product other)
    {
        Console.WriteLine($"\nПримерка: {Name} (размер {Size}) вместе с {other.Name}");
    }

    public void Wash()
    {
        Console.WriteLine($"{Name} постиран. Состав: {Material}.");
    }

    public bool IsSizeAvailable(string size)
    {
        return Size == size;
    }

    void IReturnable.ReturnProduct()
    {
        Console.WriteLine($"{Name} возвращён в магазин. Товар проверен и снова на витрине.");
    }

    public void ApplySaleDiscount()
    {
        if (IsOnSale)
        {
            Console.WriteLine($"{Name} участвует в распродаже!");
            Discount(20);
        }
        else
        {
            Console.WriteLine($"{Name} не участвует в распродаже.");
        }
    }

    public void PrintSeason()
    {
        Console.WriteLine($"{Name} — коллекция сезона «{Season}»");
    }

    public void PrintCareLabel()
    {
        Console.WriteLine($"Бирка {Name}: страна {CountryOfOrigin}, стирка при {WashTemperature}°C.");
    }

    public bool CheckGenderFit(string targetGender)
    {
        return Gender == targetGender || Gender == "Унисекс";
    }

    public void ApplySeasonalMarkup(double percent)
    {
        Price += Price * (percent / 100.0);
        Console.WriteLine($"Сезонная наценка {percent}% на {Name}. Новая цена: {Price} руб.");
    }
}

class Books : Product
{
    private string _author;
    private int _pageCount;
    private int _year;
    private string _genre;
    private string _isbn;
    private string _coverType;
    private string _language;
    private int _ageRestriction;

    public string Author
    {
        get { return _author; }
        set { _author = value; }
    }

    public int PageCount
    {
        get { return _pageCount; }
        set { _pageCount = value; }
    }

    public int Year
    {
        get { return _year; }
        set { _year = value; }
    }

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public string ISBN
    {
        get { return _isbn; }
        set { _isbn = value; }
    }

    public string CoverType
    {
        get { return _coverType; }
        set { _coverType = value; }
    }

    public string Language
    {
        get { return _language; }
        set { _language = value; }
    }

    public int AgeRestriction
    {
        get { return _ageRestriction; }
        set { _ageRestriction = value; }
    }

    public Books(string name, double price, string manufacturer, int stock,
                 string author, int pageCount, int year)
        : base(name, price, manufacturer, stock)
    {
        Author = author;
        PageCount = pageCount;
        Year = year;
    }

    public override string GetInfo()
    {
        return $"Название: {Name}, Автор: {Author}, Цена: {Price} руб., Издатель: {Manufacturer}, Страниц: {PageCount}, Год: {Year}";
    }

    public void ReadTogether(Product other)
    {
        Console.WriteLine($"\nЧтение: {Name} ({Author}) и {other.Name}");
    }

    public void Read(int pages)
    {
        Console.WriteLine($"Прочитано {pages} страниц из {PageCount}. Осталось : {PageCount - pages}.");
    }

    public bool IsBestseller()
    {
        return Year >= 2020;
    }

    public void PrintBookCard()
    {
        Console.WriteLine($"Карточка книги: «{Name}», {Author}, жанр: {Genre}, ISBN: {ISBN}");
    }

    public double DaysToRead(int pagesPerDay)
    {
        double days = Math.Ceiling((double)PageCount / pagesPerDay);
        Console.WriteLine($"«{Name}» ({PageCount} стр.) при {pagesPerDay} стр/день читается ~{days} дн.");
        return days;
    }

    public void CheckAgeAccess(int readerAge)
    {
        Console.WriteLine(readerAge >= AgeRestriction
            ? $"Читатель ({readerAge} лет) допущен к книге «{Name}» ({AgeRestriction}+)."
            : $"Возрастное ограничение {AgeRestriction}+ для книги «{Name}».");
    }

    public void TranslateInfo(string targetLang)
    {
        Console.WriteLine($"Книга «{Name}» (оригинал: {Language}, {CoverType}) запрошена для перевода на {targetLang}.");
    }

    public double CalculateWeightByPages(double gramsPerPage)
    {
        double weight = PageCount * gramsPerPage;
        Console.WriteLine($"Примерный вес книги «{Name}»: {weight} г.");
        return weight;
    }
}

class ProductService
{
    private readonly IProduct _product;

    public ProductService(IProduct product)
    {
        _product = product;
    }

    public void ShowProductCard()
    {
        _product.Display();
    }

    public void ApplyPromoDiscount(double percentage)
    {
        Console.WriteLine($"Промо-скидка для {_product.Name}:");
        _product.Discount(percentage);
    }

    public void CheckAvailability()
    {
        Console.WriteLine(_product.Stock > 0
            ? $"{_product.Name} в наличии ({_product.Stock} шт.)"
            : $"{_product.Name} отсутствует на складе!");
    }
}

class ProductCollection<T> where T : Product
{
    private List<T> _products = new List<T>();

    public void Add(T product) => _products.Add(product);
    public bool Remove(T product) => _products.Remove(product);
    public int Count => _products.Count;

    public void DisplayAll()
    {
        foreach (var p in _products)
        {
            p.Display();
        }
    }

    public void DiscountAll(double percentage)
    {
        foreach (var p in _products)
        {
            p.Discount(percentage);
        }
    }
}
```
